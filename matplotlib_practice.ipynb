```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
Data={'Year' : [1920,1930,1940,1950,1960,1970,1980,1990,2000,2010,2020],
      'Exchange Rate' :[65,69,71,64,62,59,72,71,75,78,81]}
```

```python
print(Data)
print(type(Data))
```

```python
df=pd.DataFrame(Data)
print(df)
```

```python
type(df)
```

```python
df.plot(x='Year',y='Exchange Rate',kind='bar')
plt.show()
```

```python
#horizontal Bar chart
df.plot(x='Year',y='Exchange Rate',kind='barh')
plt.show()
```

```python
df.plot(x='Year',y='Exchange Rate',kind='line')
plt.show()
```

```python
df.plot(x='Year',y='Exchange Rate',kind='area')
plt.show()
```

#Using Dataset

```python
churn_df=pd.read_csv('Churn_Modelling.csv')
```

```python
churn_df.head(5)
```

```python
churn_df.Geography.value_counts()
```

```python
churn_df.Geography.value_counts().plot(kind='barh') #bar for vertical and barh or horizontal
```

```python
churn_df['Exited'].value_counts().plot(kind='bar')
```

```python
plt.bar(churn_df['Geography'], churn_df['EstimatedSalary'].mean(), color="green", edgecolor="black")
plt.title('Geography vs Estimated salary', fontdict={"fontsize" : 20,"color":"red"})
plt.show()
```

```python
plt.scatter(churn_df['Age'],churn_df['Tenure'])
plt.title('Scatter Plot')
plt.xlabel("This is X-axis.")
plt.ylabel("This is Y-axis.")
plt.show()
```
